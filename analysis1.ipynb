```python
import pandas as pd 
import numpy as np 
import openpyxl
import re 
```

```python
df = pd.read_excel("/Users/aryanbhardwaj78/Desktop/SEM 1 - 2026/DATA 7001 Intro to Data Science/Project - student mental health/data7001_group15_project/raw_data/Dataset for Analyzing University Students Behaviour.xlsx")
df.head(5)
```

```python
df.columns = df.columns.str.strip()
```

```python
df = df.rename(columns={
    df.columns[0]: 'Semester',
    df.columns[1]: 'Age(in years)',
    df.columns[2]: 'Height(in ft)',
    df.columns[3]: 'Weight(in kg)'
})
```

```python
#function to clean semester column
def clean_semester(val):
    if pd.isna(val):
        return np.nan
    
    val = str(val).lower().strip()
    
    match = re.search(r'(\d+)', val)
    if match:
        return int(match.group(1))
    
    return np.nan

df['Semester'] = df['Semester'].apply(clean_semester).astype('Int64')
```

```python
df[df['Semester'].isna()]
```

```python
#there are 4 na values in semester.
#we will impute the most occuring value in place of na
highest_occuring_sem = df['Semester'].mode()[0]
highest_occuring_sem
```

```python
df['Semester'] = df['Semester'].fillna(highest_occuring_sem)
```

```python
#Clean age column
df['Age(in years)'] = pd.to_numeric(df['Age(in years)'], errors='coerce')
df['Age(in years)'] = df['Age(in years)'].round()
df['Age(in years)'] = df['Age(in years)'].astype('Int64')
```

```python
#Imputing na values in age column with mean(age)
mean_age = df['Age(in years)'].mean()
df['Age(in years)'] = df['Age(in years)'].fillna(round(mean_age))
```

```python
#clean height column
import numpy as np
import re

def clean_height(val):
    if pd.isna(val):
        return np.nan


    if isinstance(val, (int, float)):
        return float(val)

    val = str(val).lower().strip()


    val = val.replace('"', '')
    val = val.replace("''", '')
    val = val.replace("'", ' ')
    val = re.sub(r'\s+', ' ', val).strip()

  
    if re.fullmatch(r'\d+\.\d+', val):
        return float(val)

    match = re.fullmatch(r'(\d+)\s+(\d+)', val)
    if match:
        feet = match.group(1)
        inches = match.group(2)
        return float(f"{feet}.{inches}")


    match = re.search(r'(\d+)\s*feet?\s*(\d+)', val)
    if match:
        feet = match.group(1)
        inches = match.group(2)
        return float(f"{feet}.{inches}")

    match = re.search(r'(\d+)\s*feet?', val)
    if match:
        return float(match.group(1))


    if re.fullmatch(r'\d+', val):
        return float(val)

    return np.nan
    
```

```python
df['Height(in ft)'] = df['Height(in ft)'].apply(clean_height)
```

```python
#imputing na with mean height
mean_height = df['Height(in ft)'].mean()
df['Height(in ft)'] = df['Height(in ft)'].fillna(mean_height)
```

```python
df['Height(in ft)'] = df['Height(in ft)'].round(1)
```

```python
#Cleaning weight column
def clean_weight(val):
    if pd.isna(val):
        return np.nan

    if isinstance(val, (int, float)):
        return int(round(float(val)))

    val = str(val).lower().strip()

    match = re.search(r'\d+(\.\d+)?', val)
    if match:
        return int(round(float(match.group())))

    return np.nan
```

```python
df['Weight(in kg)'] = df['Weight(in kg)'].apply(clean_weight)
```
